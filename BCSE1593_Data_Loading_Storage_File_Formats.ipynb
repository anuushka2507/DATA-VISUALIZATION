## 2. File formats in data science

Different types of files are commonly used for storing data.

| Format | Common extension | Typical use |
|---|---|---|
| CSV | `.csv` | tabular data, very common |
| TSV / delimited text | `.tsv`, `.txt` | tabular text with another separator |
| JSON | `.json` | structured and nested data |
| Excel | `.xlsx` | spreadsheets and business data |
| Pickle | `.pkl` | storing Python objects |
| Parquet | `.parquet` | efficient analytical data storage |

There is no single "best" format for every situation.

For example:

- CSV is simple and widely supported.
- Excel is convenient for spreadsheet users.
- JSON is useful for structured records and nested data.
- Pickle can preserve Python objects.
- Parquet is designed for efficient analytical workloads.

In this practical, we will focus on formats that can be used comfortably with pandas.

## 6. Why CSV?

CSV stands for **Comma-Separated Values**.

A simple CSV file may look like:

```text
Name,Department,Salary
Aman,IT,55000
Priya,HR,48000
Kunal,Sales,62000
```

CSV is a text format.

It is one of the most commonly used formats for tabular data because it is:

- simple
- readable
- supported by many tools
- easy to exchange between systems

Pandas provides `read_csv()` for loading CSV data.

## 11. Useful `read_csv()` options

`read_csv()` has many options.

Some useful ones include:

| Option | Purpose |
|---|---|
| `sep` | specify the separator |
| `header` | identify the header row |
| `names` | provide column names |
| `index_col` | use a column as the index |
| `usecols` | load selected columns |
| `nrows` | load a limited number of rows |
| `skiprows` | skip rows |

You do not need to memorise every option. The important skill is knowing that the function can be adjusted when a file does not follow the default CSV structure.

# Part E – JSON Files

## 21. Understanding JSON

JSON stands for **JavaScript Object Notation**.

It is a text format commonly used for semi-structured data.

A simple JSON file may look like:

```json
[
    {"Name": "Aman", "Department": "IT", "Salary": 55000},
    {"Name": "Priya", "Department": "HR", "Salary": 48000}
]
```

JSON is especially useful when data has a nested or record-based structure.

Pandas provides `read_json()` for many common JSON structures.

# Part H – Binary Storage with Pickle

## 32. What is a binary format?

Text formats such as CSV and JSON store information as readable text.

Binary formats store information in a form designed for computers rather than direct human reading.

One Python-specific binary format is **Pickle**.

Pandas can save a DataFrame using `.to_pickle()` and load it using `read_pickle()`.

Pickle can preserve Python objects more directly than CSV.

### Important note

Pickle files should only be loaded from trusted sources. Do not load arbitrary pickle files from unknown websites.

## 33. Saving a DataFrame as Pickle

### Live coding

Save `employees_demo` as:

```text
employees_demo.pkl
```

Then load it into `employees_pickle`.

Display the loaded DataFrame.

````python
print("Saving as Pickle:")

# Save employees_demo as a pickle file
employees_demo.to_pickle("employees_demo.pkl")
print("Saved.")

# Read the pickle file
print("Reading Pickle:")
employees_pickle = pd.read_pickle("employees_demo.pkl")
print(employees_pickle)


# Display the loaded DataFrame
print(employees_pickle)
````

## 34. Comparing CSV and Pickle

Consider the same DataFrame stored as CSV and Pickle.

### CSV

- text based
- easy to inspect
- widely supported
- good for data exchange

### Pickle

- binary
- Python-friendly
- can preserve Python objects
- not intended as a universal exchange format
- should only be loaded from trusted sources

The choice depends on the purpose of the file.

# Part I – Modern Analytical File Format

## 35. Parquet

**Parquet** is a column-oriented file format commonly used in data engineering and analytics.

It is designed for efficient storage and reading of tabular data.

Compared with CSV, Parquet can:

- store data types more efficiently
- support compression
- read selected columns efficiently
- work well with large analytical datasets

Pandas can work with Parquet with a suitable engine such as `pyarrow`.

### Live coding

Save `employees_demo` as `employees_demo.parquet` and read it back.

````python
# Check whether pyarrow is available.
# If needed, install it in Colab before running the next commands.

print("Parquet practice:")

# Save employees_demo as Parquet
employees_demo.to_parquet("employees_demo.parquet")
print("Saved.")

# Read the Parquet file
employees_parquet = pd.read_parquet("employees_demo.parquet")
# Read employees_demo.parquet back into pandas
print(employees_parquet)
````
